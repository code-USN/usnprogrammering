## Nullpunkter
Her er et forslag til pseudokode for å finne nullpunkter i en andregradsfunksjon:
1. Lag 3 variabler for a, b og c som i det generelle andregradsuttrykket $ax^2+bx+c$
2. Bruk formelen for symmetrilinjen til andregradsfunksjonen og finn ekstremalpunktet: $x=-\frac{b}{2a}$
3. Bruk fortegnet til a til å bestemme om dette er et topp- eller bunnpunkt, og skriv ut koordinatet på formen (x,y)
4. Sjekk om funksjonen faktisk har nullpunkter (er $b^2 < 4ac$ ?). Andregradfunksjoner har enten 0, 1 eller 2 nullpunkter!
5. Hvis den har nullpunkter, bruk løsningsformelen: $x = \frac{-b \pm \sqrt{b^2-4ac}}{2a}$ til å finne nullpunktene

Eksempel til andregradsfunksjon $f(x)=x^2-x-12$

# nødvendig bibliotek for å kunne ta kvadrat
import math as m

# Eksempelet undersøker f(x)=x^2-4
# Dere kan endre a, b og c for å undersøke andre funksjoner!
a = 1
b = 0
c = -4

# Finner nullpunkter hvis de eksisterer
if b**2 < 4*a*c:
    print("Funksjoner har ingen nullpunkter")
elif b**2 == 4*a*c:
    print("Funksjonen har ett nullpunkt:")
    x = (-b + m.sqrt(b**2-4*a*c))/(2*a)
    print("x = " + str(x))
else:
    print("Funksjonen har to nullpunkter:")
    x = (-b - m.sqrt(b**2-4*a*c))/(2*a)
    print("x1 = " + str(x))
    x = (-b + m.sqrt(b**2-4*a*c))/(2*a)
    print("x2 = " + str(x))

# Finner symmetrilinjen
x = (-b)/(2*a)
y = a*x**2+b*x+c
if a < 0:
    print("Funksjonen har et toppunkt for x = " + str(x) + ", y = " + str(y))
else:
    print("Funksjonen har et bunnpunkt for x = " + str(x) + ", y = " + str(y))
