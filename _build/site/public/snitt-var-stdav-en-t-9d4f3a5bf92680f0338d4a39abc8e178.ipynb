# Gjennomsnitt, varians og standardavvik for EN terning

```{code} python
# Importerer bibliotek for tilfeldighet
import random as rnd

# Antall sider på terninger
sider = 6

# Antall forsøk
antall_forsøk = 10

forventning = (sider+1)/2

# variabler for å huske summer i sløyfen
variansteller = 0
summen = 0

for i in range(antall_forsøk):
    terning = rnd.randint(1, sider)
    print("kast " + str(i+1) + ": " + str(terning) ) #bør fjernes veh høyt antall
    variansteller = variansteller + (terning - forventning)**2
    summen = summen + terning

# Her kan du lage kode for å regne ut:
# * gjennomsnitt
# * varians og standardavvik
# * skriv ut resultatet med print

# Eksempel på kode for utskrift
print("Summen er:", summen)
```
