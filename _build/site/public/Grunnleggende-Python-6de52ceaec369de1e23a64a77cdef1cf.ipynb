## Variabler og skrive ut til skjerm (print)

Programmet nedenfor lager en tallvariabel og en tekstvariabel og skriver ut innholdet på forskjellige måter. Hvordan tror du det fungerer? (PS Linjene som begynner med # er kommentarer til koden ment for å forklare underveis, disse linjene er IKKE en del av programmet!). Resultatet av programmet ser du nedenfor koden.

```{note} Bruk av = i programmering
Vær obs på at likhetstegnet = ikke er helt det samme i programmering som i matematikk. I programmering bruker man = som en tildelingsoperator. Dette er viktig å kunne, siden det kan forsterke en vanlig misoppfatning om likhetstegnet i matematikk. Se lenger ned på siden for hvordan man bruker ekvivalens i programmering.
```

````python
#vi skriver inn en verdi for x 
x = 10 
#vi skriver ut verdien til x  
print(x) 
#tar med tekst i tillegg til verdien av x 
print("x er lik", x) 
#skriver ut navnet ditt 
print("Ola Nordmann") 
#Definerer ditt navn til å være en variabel vi kaller dittnavn 
dittnavn = "Ola Nordmann" 
#skriver ut første bokstav av navnet ditt 
print(dittnavn[0]) 
#Tar ut deler av navnet fra bokstav 4 til 12 
print(dittnavn[4:12])
````

## Regneoperasjoner

Vi lager et nytt program, her skal vi undersøke hvordan vi kan addere, subtrahere, multiplisere og dividere med operatorer i Python.

Vi:
* Definerer to variabler a og b
* Vi adderer disse og skriver ut svaret
* Vi gjentar dette for subtraksjon, multiplikasjon og divisjon

````python
#definerer variablene a og b 
a = 4 
b = 8 
 
#summerer a og b og skriver ut summen 
sum = a + b 
print("Summen er",sum) 
 
#subtraherer a fra b og skriver ut summen 
sub = b - a 
print("Differansen er",sub) 
 
#dividerer a og b og skriver ut summen 
div = a/b 
print("Kvotient er", div) 
 
#multipliserer a og b og skriver ut summen 
mult = a*b 
print("Produktet er", mult)
````

```{tip} Oppgave
Kopier denne koden til din egen Python-editor, og sjekk selv. Prøv å endre på verdiene av variablene a og b og se hva som skjer!
```

## Hente inn verdier fra brukeren

I eksemplene over er verdiene til variablene definert i selve koden. Det går også an å la den som bruker programmet skrive inn verdier når koden kjøres. I eksempelet under lar vi brukeren bestemme verdiene til a og b ved hjelp av kommandoen input().

```{note} Datatyper
I Pyhton er kan variabler både være tallverdier og tekst, men python kan ikke gjøre beregninger på tekst, derfor er inputkommandoen "nøstet" inne i en int() kommando som gjør tekstverdien til svaret om til et heltall. Altså hvis bruker skriver verdien(teksten) "4" så gjør int() kommandoen verdien om til tallet 4. Vi skal se nærmere på datatyper senere!
```

````python
#importerer to valgfrie variable a og b 
a = int(input("Hvilket heltall er a?")) 
b = int(input("Hvilket heltall er b?")) 
#summerer a og b og skriver ut svaret 
summen = a + b 
print("Summen er",summen) 
#subtraherer a fra b og skriver ut svaret 
sub = b - a 
print("Differansen er",sub) 
#dividerer a og b og skriver ut svaret 
div = a/b 
print("Kvotient er", div) 
#multipliserer a og b og skriver ut svaret 
mult = a*b 
print("Produktet er", mult) 
````

```{tip} Oppgave
Kopier denne koden til din egen Python-editor, og sjekk selv. Nå kan du skrive inn forskjellige verdier for a og b hver gang koden kjøres!
```

## Utskrifter

Vi har allerede brukt utskriftsfunksjonen print() i eksemplene over. Det er mange måter å sette sammen utskrifter på, i eksemplet under skal vi se på 2 forskjellige måter. Legg merke til at den første metoden er enklest. I den andre metoden må vi endre tallverdien til div til en tekst, og vi får ikke automatisk mellomrom.

````python
# vi tar inn to verdier 
a = 5 
b = 7 

#vi dividerer verdiene og skriver ut resultatet 
div = a/b 

# Metode 1: kommaseparaert
print("divisor er lik", div)

# Metode 2: satt sammen med +
print("divisor er lik " + str(div))
````
