# Hvordan bruke nettsiden

Alle aktiviteter og eksempler på denne siden krever at man har et python-program installert på egen maskin. Vi har bevisst unnlatt å legge inn interaktive kodeceller, da disse krever kontinuerlig støtte fra 3. parts nettsider. For eksempel forsvinner en av de mest populære sidene for dette, trinket.io i august 2026.

Alle kodeeksempler på denne nettsiden kan enkelt kopieres inn i eget python-program med "copy-paste". Det finnes til og med en egen knapp øverst til høyre i alle kodecellene som kopierer koden til buffer!

Vi anbefaler følgende pythonprogrammer
* Spyder
* Thonny

## Thonny
### Fordeler
* Liten installallasjon
* Enkle menyer
* Støtter Turtle bra
* Plotter i egne vinduer som standard, lettere å få fine plots som kan kopieres til feks word
* Har enkel støtte for syntaks med farger for innebygde kommandoer
### Ulemper
* Må installere noen biblioteker manuelt (matplotlib, numpy
* Har ikke skrivestøtte (code completion)
* Variabelinspektøren er noe enkel
* Må oppdateres manuelt
* Plott i egne vinduer gjør det litt mer klumsete å sammenligne å sammenligne flere plots
## Spyder (standalone)
### Fordeler
* Kommer med de aller fleste nødvendige biblioteker forhåndsinstallert
* Meget god skrivestøtte med "code completion"
* Variabelinspektøren støtter alle variabeltyper, og lar bruker inspisere lister og arrays i egne vinduers
* Plotter i innebygd vindu som standard, raskere og lettere å sammenligne forskjellige plots
* Har mulighet for KI støtte
### Ulemper
* Større installasjon
* Støtter Turtle dårligere enn Thonny
* Litt mer komplisert meny
